```python
import sqlite3
```

```python
connection = sqlite3.connect('test1.db')
cursor = connection.cursor()

backup = sqlite3.connect('backup_test.db')

connection.backup(backup)   
```

```python
cursor.execute('''
create table if not exists employee(
    id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    age INTEGER,
    department TEXT
    )
''')
```

```python
data = [
    (18,'shiva',21,'cse'),
    (19,'sheshank',22,'csd'),
    (20,'bhavishya',23,'aiml'),
    (21,'sai',24,'cybersecurity')
]
```

```python
cursor.executemany('''
insert into employee(id,name,age,department)
values(?,?,?,?)
''',data)
connection.commit()
```

```python
cursor.execute('''
    select * from employee
''')
results = cursor.fetchall()
for result in results:
    print(result)


```
