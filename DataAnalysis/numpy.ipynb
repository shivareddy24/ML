```python
a = np.random.randint(1,20, size=(1,6))
print("original : " ,a)
mean = a.mean()
print("mean : " ,mean)
std_dev = a.std()
print("standard deviation : " ,std_dev)
median = np.median(a)
print("median : " ,median)
variance = np.var(a)
print("variance : " ,variance)
```

```python
a = np.random.randint(1,20, size=(1,6))
print("original : " ,a)
mean = a.mean()
print("mean : " ,mean)
std_dev = a.std()
print("standard deviation : " ,std_dev)

normalized = (a-mean)/std_dev
print("normalized : " ,normalized)
```
