```python
df = pd.DataFrame(np.random.randint(1, 100, size=(5, 3)), columns=['A', 'B', 'C'])
print("Original DataFrame:")
print(df)
df.iloc[0,2]=np.nan
df.iloc[1,1]=np.nan
df.iloc[3,0]=np.nan
print("\nDataFrame after introducing NaN values:")
print(df)
df.fillna(df.mean(),inplace=True)
print("\nDataFrame after filling NaN values with column means:")
print(df)
```

dropna = it uses to drop the NaN column

fillna = it uses to fill the NaN value 

fillna : it uses to replace the NaN value 

inplace = True : it will replace the excisted dataframe instead of creating new datafarame

```python
import pandas as pd
import numpy as np
df=pd.DataFrame(np.random.randint(1,100,size=(6,5)) ,columns=['A','B','C','D','E'])
print("Original DataFrame:")
print(df)

df.iloc[0,2]=np.nan
df.iloc[1,1]=np.nan
df.iloc[3,3]=np.nan
print("\nDataFrame after introducing NaN values:")
print(df)


df.dropna(inplace=True)
print("\nDataFrame after dropping rows with NaN values:")
print(df)
```
