```python
df = pd.DataFrame(np.random.randint(1, 100, size=(5, 3)), columns=['A', 'B', 'C'])
print("Original DataFrame:")
print(df)
df.iloc[0,2]=np.nan
df.iloc[1,1]=np.nan
df.iloc[3,0]=np.nan
print("\nDataFrame after introducing NaN values:")
print(df)
df.fillna(df.mean(),inplace=True)
print("\nDataFrame after filling NaN values with column means:")
print(df)
```

```python
import pandas as pd
import numpy as np
df=pd.DataFrame(np.random.randint(1,100,size=(6,5)) ,columns=['A','B','C','D','E'])
print("Original DataFrame:")
print(df)

df.iloc[0,2]=np.nan
df.iloc[1,1]=np.nan
df.iloc[3,3]=np.nan
print("\nDataFrame after introducing NaN values:")
print(df)


df.dropna(inplace=True)
print("\nDataFrame after dropping rows with NaN values:")
print(df)
```

groupby : when we create multiple repeated values in a column it separate them by single value

          groupby('--name--')['-columns to perform-']

aggregate : it performs step by step operation on selected columns (generally for multiple operations ex: sum,mean,count)


```python
import pandas as pd
import numpy as np
df = pd.DataFrame(
    {
        'category': np.random.choice(['A','B','C'], size=10),
        'value': np.random.randint(1, 100, size=10)
    })
print("Original DataFrame:")
print(df)

df.groupby('category')['value'].agg(['mean', 'sum', 'count'])
```

```python
import pandas as pd
import numpy as np  
df = pd.DataFrame(
    {
        'product': np.random.choice(['Product A', 'Product B', 'Product C'], size=10),
        'category': np.random.choice(['A','B','C'], size=10),
        'sales': np.random.randint(1,100,size=10)
    }
)
print("Original DataFrame:")
print(df)

df.groupby(['product','category'])['sales'].sum()
```
