groupby : when we create multiple repeated values in a column it separate them by single value

          groupby('--name--')['-columns to perform-']

aggregate : it performs step by step operation on selected columns (generally for multiple operations ex: sum,mean,count)


```python
import pandas as pd
import numpy as np
df = pd.DataFrame(
    {
        'category': np.random.choice(['A','B','C'], size=10),
        'value': np.random.randint(1, 100, size=10)
    })
print("Original DataFrame:")
print(df)

df.groupby('category')['value'].agg(['mean', 'sum', 'count'])
```

```python
import pandas as pd
import numpy as np  
df = pd.DataFrame(
    {
        'product': np.random.choice(['Product A', 'Product B', 'Product C'], size=10),
        'category': np.random.choice(['A','B','C'], size=10),
        'sales': np.random.randint(1,100,size=10)
    }
)
print("Original DataFrame:")
print(df)

df.groupby(['product','category'])['sales'].sum()
```

merge : it merges to dataframes by single column

concatenate : it adds another dataframe into one by row or column


```python
#merge

import pandas as pd 
import numpy as np

df1 = pd.DataFrame(
    {
        'keys' : ['A', 'B', 'C', 'D', 'E'],
        'values' : np.random.randint(1, 100, size=5)
    }
)
df2 = pd.DataFrame(
    {
        'keys' : ['A', 'B', 'C', 'D', 'E'],
        'values' : np.random.randint(1, 100, size=5)
    }
)
print("DataFrame 1:")
print(df1)          
print("\nDataFrame 2:")
print(df2)  

merge = df1.merge(df2, on='keys')
print("\nMerged DataFrame:")    
print(merge)
```

```python
#concat with row

import pandas as pd 
import numpy as np

df1 = pd.DataFrame(
    {
        'keys' : ['A', 'B', 'C', 'D', 'E'],
        'values' : np.random.randint(1, 100, size=5)
    }
)
df2 = pd.DataFrame(
    {
        'keys' : ['x', 'y', 'z', 'w'],
        'values' : np.random.randint(1, 100, size=4)
    }
)
print("DataFrame 1:")
print(df1)          
print("\nDataFrame 2:")
print(df2)  

rows = pd.concat([df1,df2],axis=0)
print("\nConcatenated DataFrame (by rows):")
print(rows) 
```
