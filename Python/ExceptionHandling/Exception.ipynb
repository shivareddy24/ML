```python
# zerodivision exception
def div(x,y):
    try:
        a = x/y
        print(a)
    except ZeroDivisionError as e:
        print(f"it forms a {e}. ")
    finally:
        print("Execution Done.")

div(4,5)
div(7,0)

```

```python
def sum(lst):
    total = 0
    try:
        for i in lst:
            total += i
        print(total)
    except TypeError as e:
        print(f"it shows {e}")
    finally:
        print("Execution Done.")
lst=[1,2,3,4,5,'r']
sum(lst)
```

```python
def dictionary(d,key):
    try:
        val = d[key]
        print(val)
    except KeyError as ex:
        print(f"the error : {ex}")
    finally:
        print("execution done.")
dict={'a':1,'b':2,'c':3}
dictionary(dict,'a')
dictionary(dict,'d')
```

```python
def number(n):
    try:
        try:
            num = int(n)
        except ValueError as e:
            print(f"error is : {e}")
            num = None
        finally:
            print("-----------")

        if num is not None:
            try:
                x = 10 / num
                print(x)
            except ZeroDivisionError as e:
                print(f"error is {e}")
            finally:
                print("execution done.")

        return num

    finally:
        print("overall execution done.")

print(number(5))
   
```

```python
def display(lst,n):
    try:
        num=lst[n]
        print(num)
    except IndexError as e:
        print(f"error : {e}")
    finally:
        print("execution done")

list = [1,2,3,5,6,7,8]
display(list,3)
display(list,13)


```

```python
class NegativeNumberError(Exception):
    pass

def negative(lst):

    try:
        for i in lst:
            if i < 0:
                raise NegativeNumberError(f"negative number found {i}.")
    except NegativeNumberError as e:
        print(f"error : {e}")
    finally:
        print("execution done.")

list = [1,2,3,4,5,6,5]
list1 = [1,2,3,4,3,2,-1]
negative(list)
negative(list1)
```
