```python
# zerodivision exception
def div(x,y):
    try:
        a = x/y
        print(a)
    except ZeroDivisionError as e:
        print(f"it forms a {e}. ")
    finally:
        print("Execution Done.")

div(4,5)
div(7,0)

```

```python
def sum(lst):
    total = 0
    try:
        for i in lst:
            total += i
        print(total)
    except TypeError as e:
        print(f"it shows {e}")
    finally:
        print("Execution Done.")
lst=[1,2,3,4,5,'r']
sum(lst)
```
