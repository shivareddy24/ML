```python
from abc import ABC,abstractmethod
class Vehicle(ABC):

    @abstractmethod
    def engine_start(self):
        pass

class Car(Vehicle):
    def engine_start(self):
        return " car engine started"

class Bike(Vehicle):
    def engine_start(self):
        return "bike engine started"

car=Car()
bike=Bike()
print(car.engine_start())
```

```python
class Vehicle(ABC):
    @abstractmethod
    def start_engine(self):
        pass

    def fuel_type(self):
        return "Generic Fuel"

class Car(Vehicle):
    def start_engine(self):
        print("Car engine started")

    def fuel_type(self):
        return "diesel"

class Bike(Vehicle):
    def start_engine(self):
        print("Bike engine started")

    def fuel_type(self):
        return "Petrol"

car=Car()
bike=Bike()
print(bike.fuel_type())
print(car.fuel_type())
```

```python
class BankAccount:
    def __init__(self,accno,balance):
        self.__accno=accno
        self.__balance=balance
    def deposit(self,amount):
        self.__balance += amount
    def withdraw(self,amount):
        if self.__balance > amount:
            self.__balance -= amount
        else:
            print("not possible")
    def check_balance(self):
        return self.__balance

ba=BankAccount(2411,50000)
ba.deposit(500)
ba.withdraw(8700)
print(ba.check_balance())
```
