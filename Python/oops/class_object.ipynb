```python
class BankAccount:
    def __init__(self,accno,balance):
        self.__accno=accno
        self.__balance=balance
    def deposit(self,amount):
        self.__balance += amount
    def withdraw(self,amount):
        if self.__balance > amount:
            self.__balance -= amount
        else:
            print("not possible")

ba=BankAccount(2411,50000)
ba.deposit(500)
ba.withdraw(8700)
        
```

```python
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age

class Employee(Person):
    def __init__(self,name,age,emp_id):
        super().__init__(name,age)
        self.emp_id=emp_id

    def __str__(self):
        return f"Employee(name : {self.name} ,age : {self.age} ,employee_id : {self.emp_id})"

emp = Employee("shiva reddy",30,1234)
print(emp)
```
