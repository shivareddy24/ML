```python
#fibonacci
def fibonacci(n):
    t1 = 0
    t2 = 1
    for i in range(n):
        print(t1,end=" ")
        temp = t1
        t1 = t2
        t2 += temp
n = int(input("Enter the number of terms: "))
fibonacci(n)
```

```python
def add_to_dict(a, b=None):
    if b is None:
        b = {}
    b[a] = a**2
    return b

print(add_to_dict(2)) 
print(add_to_dict(3, {1: 1}))
```

```python
def filter_integers(**kwargs):
    return {k: v for k, v in kwargs.items() if isinstance(v, int)}

print(filter_integers(a=1, b='two', c=3, d=4.5))
print(filter_integers(x=10, y='yes', z=20))
```

```python
def add(x,y):
    return x+y
def mul(x,y):
    return x*y  

def calculate(operation,x,y):
    print(operation(x,y))

calculate(add,4,5)
calculate(mul,8,7)
```

```python
def outer(n):
    def square(n):
        return n**2
    return square(n)
print(outer(8))
```

```python
def decorate(func):
    def inner():
        name = ["shiva" , "prasad" , "reddy"]
        for i in name:
            func(i)
    return inner
@decorate
def display(s):
    print(s.upper())

display()     
```
