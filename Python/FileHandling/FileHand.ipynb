```python
#write a file
with open("example.txt",'w') as file:
    file.write("kathi shivaaaaa reddy")
```

```python
#read a file
with open("example.txt",'r') as file:
    text = file.read()
    print(text)
```

```python
with open("example.txt",'r') as file:
    text = file.read()

with open('ex.txt','w') as dest:
    dest.write(text)

print(text)
```

```python
with open('log.txt','w') as file:
    file.write("hello everyone! \n welcome")

with open('log.txt','a') as fl:
    fl.write("\nhow are you?")
```

```python
#counting words and letters
with open('example.txt','r') as file:
    text = file.read()
words = len(text.split())
letters = len(text)

print(words)
print(letters)
```

```python
#finding and replacing
with open('example.txt','r') as file:
    text = file.read()
    text = text.replace("shivaaaaa","shiva")
print(text)
```

```python
with open("example.txt", "r") as file:
    lines = file.readlines()

for line in lines:
    print(line.strip()[::-1])
```

```python
with open('example.txt','r') as file:
    text = file.read()
words = len(text.split())
letters = len(text)
lines = len(text.splitlines())

print(words)
print(letters)
print(lines)
```

```python
with open('example.txt','r') as file:
    text = file.read()
with open('log.txt','r') as source:
    t = source.read()

with open('output.txt','w') as out:
    out.write(text)
    out.write(t)
    
```

```python
with open("example.txt", "r") as file:
    lines = file.readlines()

for i, line in enumerate(lines, start=1):
    with open(f"file{i}.txt", "w") as out:
        out.write(line)
```
