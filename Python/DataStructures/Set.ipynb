```python
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

u = s1.union(s2)
print(f"union : {u}")
i = s1.intersection(s2)
print(f"intersection : {i}")
d = s1.difference(s2)
print(f"difference : {d}")
```

```python
nums = {1,2,3,4,5,6,7,8,9}
square = {i**2 for i in nums}
print(f"Original set: {nums}")
print(f"Square set: {square}")  
```

```python
nums = {1,2,3,4,5,6,7,8,9}
even = set()
for i in nums:
    if i%2 == 0 :
        even.add(i)

print(f"Original set: {nums}")
print(f"Even set: {even}")
```

```python
s = frozenset([1,2,3,4,5])
print(type(s))
print(s)
```

```python
set1 = {1,2,3,4,5}
set2 = {4,5}
print(f"Is set2 a subset of set1? {set2.issubset(set1)}")
print(f"Is set1 a superset of set2? {set1.issuperset(set2)}")
```
