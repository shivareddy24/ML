```python
def transpose(matrix):
    trans = []
    for i in range(len(matrix[0])):
        row = []
        for j in range(len(matrix)):
            row.append(matrix[j][i])
        trans.append(row)
    return trans

matrix = [
    [0, 23, 65],
    [6, 5, 6],
    [73, 4, 9]
]
transposed = transpose(matrix)
print("Original matrix:")
for row in matrix:
    print(row)
print("Transposed matrix:")
for row in transposed:
    print(row)
```
