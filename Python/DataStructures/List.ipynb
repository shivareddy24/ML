```python
a=list(range(1,20))
print(f"List Starting element is : {a[0]}")
print(f"List middle element is : {a[len(a)//2]}")
print(f"List Ending element is : {a[-1]}")
```

```python
a=list(range(1,20))
print(f"List Starting 5 elements are : {a[:5]}")
print(f"List middle 5 elements are : {a[5:15]}")
print(f"List Ending 5 elements are : {a[-5:]}")
```

```python
num = list(range(1,10))
square =[]
for i in num:
    square.append(i**2)

print(f"numbers are :{num}")
print(f"squares are :{square}") 
```

```python
num = list(range(1,20))
even =[]
for i in num:
    if i%2==0:
        even.append(i)

print(f"numbers are :{num}")
print(f"after filtering the even numbers are :{even}") 
```

```python
num = [65,12,68,45,68,67,234,12,34,56,78,90]
x = sorted(set(num))
print(f"numbers after removing duplicates and in order are :{x}")
print(f"numbets after removing duplicates and in reverse order are :{x[::-1]}")
```
