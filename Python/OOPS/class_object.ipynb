```python
class BankAccount:
    def __init__(self,accno,balance):
        self.__accno=accno
        self.__balance=balance
    def deposit(self,amount):
        self.__balance += amount
    def withdraw(self,amount):
        if self.__balance > amount:
            self.__balance -= amount
        else:
            print("not possible")

ba=BankAccount(2411,50000)
ba.deposit(500)
ba.withdraw(8700)
        
```

```python
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age

class Employee(Person):
    def __init__(self,name,age,emp_id):
        super().__init__(name,age)
        self.emp_id=emp_id

    def __str__(self):
        return f"Employee(name : {self.name} ,age : {self.age} ,employee_id : {self.emp_id})"

emp = Employee("shiva reddy",30,1234)
print(emp)
```

```python
class Address:
    def __init__(self,street,city,zipcode):
        self.street=street
        self.city=city
        self.zipcode=zipcode

class Person(Address):
    def __init__(self,name,age,address):
        self.name=name
        self.age=age
        self.address=address

address=Address('teachers colony','hyderabad',501510)
person=Person('shivareddy',20,address)
print(f"the person name {person.name} of age {person.age} , live in {address.street,address.city,address.zipcode}")
```

```python
class Rectangle:
    def __init__(self, length, width):
        self.__length = length
        self.__width = width

    @property
    def length(self):
        return self.__length

    @length.setter
    def length(self, length):
        self.__length = length

    @property
    def width(self):
        return self.__width

    @width.setter
    def width(self, width):
        self.__width = width

rect = Rectangle(10, 5)
print(rect.length, rect.width) 
rect.length = 15
rect.width = 7
print(rect.length, rect.width)
```

```python
from abc import ABC,abstractmethod
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass
class Square(Shape):
    def area(self,x):
        self.x=x
        return x**2
class Circle(Shape):
    def area(self,r):
        self.r=r
        return 3.14*r*r
def display(shape,x):
    print(shape.area(x))

square=Square()
circle=Circle()
display(square,7)
display(circle,5)


```

```python
class Vector:
    def __init__(self,x,y):
        self.x=x
        self.y=y

    def __add__(self,other):
        return self.x + other.x , self.y + other.y

v1=Vector(2,3)
v2=Vector(5,6)
print(v1+v2)
```
