```python
dict = {'shiva' : 1, 'bhavishya' : 2, 'siddharth' : 3, 'sachin' : 4}
print(dict)
print(dict['bhavishya'])
print(dict.get('shiva'))
print(dict.keys())
print(dict.values())
```

```python
dict = {i: i**2 for i in range(1,10)}
print(dict)
```

```python
#accessing a value by its key
print(dict[5])
```

```python
# adding a new key-value pair

dict[11] = 121
print(dict)
```

```python
#iterating through a dictionary

for k,v in dict.items():
    print(f"{k} : {v}")
```

```python
# Merge two dictionaries

dict1 = {i : i**2 for i in range(1,5)}
dict2 = {i : i**2 for i in range(5,10)}
dict1.update(dict2)
print(dict1)
```

```python
student = {
    'name' : 'shiva',
    'age' : 22,
    'grade' : {
        'maths' : 90,
        'physics' : 80,
        'chemistry' : 95
    }
}
print(student)
```

```python
from collections import defaultdict

grade = defaultdict(list)
grade['maths'].append(90)
grade['physics'].append(80)
grade['chemistry'].append(95)   
print(grade)
```

```python
d = {i: i**2 for i in range(1, 10)}
print(d)
swap = {v: k for k,v in d.items()}
print(swap)
```
