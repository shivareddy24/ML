```python
class Car:
    def __init__(self,name,model,year):
        self.name=name
        self.model=model
        self.year=year
    def car(self):
        print(f"name of the car is {self.name} of model {self.model} of year {self.year}")

costumer = Car("-x-","-y-","-z-")
costumer.car()
```

```python
class BankAccount:
    def __init__(self,owner,balance=0):
        self.owner=owner
        self.balance=balance
    
    def deposit(self,amt):
        self.balance+=amt
        print(f"{amt} is deposited. new balance is {self.balance}")

    def withdraw(self,amt):
        if amt>self.balance :
            print("insufficiant funds : ")
        else:
            self.balance-=amt
            print(f"{amt} is withdrawed . new amount is {self.balance}")
    
    def get_balance(self):
        return self.balance
    
account=BankAccount("shivareddy",10000)
print(account.balance)
account.deposit(400)
account.withdraw(500)
print(account.get_balance())
```

```python
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age

class Employee(Person):
    def __init__(self,name,age,emp_id):
        super().__init__(name,age)
        self.emp_id=emp_id

emp=Employee("shivkha",20,18)
print(emp.name)
    
```
