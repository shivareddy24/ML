```python
import matplotlib.pyplot as plt
x = [1, 2, 3, 4, 5]
y = [1, 4, 9, 16, 25]
plt.plot(x,y)
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [1, 4, 9, 16, 25]
plt.plot(x,y,color='red',linestyle='--',marker='o',markerfacecolor='blue')
plt.title('sample graph')
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [1, 4, 9, 16, 25]
z = [1, 8, 27, 64, 125]

plt.figure(figsize=(9,5))

plt.subplot(2,2,1)
plt.plot(x,y,color='green',linestyle='--',marker='o')

plt.subplot(2,2,2)
plt.plot(x,z,color='black',linestyle='--',marker='o')

plt.subplot(2,2,3)
plt.plot(y,z,color='blue',linestyle='--',marker='o')

plt.subplot(2,2,4)
plt.plot(y,z,color='red',linestyle='--',marker='o')
```
